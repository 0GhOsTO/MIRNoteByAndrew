# Note 5: Audio Features

```python
import IPython.display as ipd
import matplotlib.pyplot as plt
import librosa.display
import numpy

import urllib.request
```

```python
url = "https://raw.githubusercontent.com/musicinformationretrieval/musicinformationretrieval.com/gh-pages/mirdotcom/assets/audio/simple_loop.wav"
fp, _ = urllib.request.urlretrieve(url, "simple_loop.wav")
x, sr = librosa.load(fp)
ipd.Audio(x, rate=sr)
```

```python
print("sr: ", sr)
librosa.display.waveshow(x, sr = sr)
plt.ylabel("Amplitude")
```

```python
# computing the short time fourier transferm(stft)
X = librosa.stft(x)
Xmag = librosa.amplitude_to_db(X)
librosa.display.specshow(Xmag, sr=sr, x_axis="time", y_axis="log")
```
