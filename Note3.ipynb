# Note 3: Tuning System


## Common MIR Features

- **Equal Temperament** - all twelve semitones and octave have the same width. $2^{1/12}$
- **Cents** - this same interval is 100 cents(octave has 1200 cents).

```python
import numpy as np
import matplotlib.pyplot as plt

#frequency
f0 = 440
# semitone steps(0 to 12)
n = np.arange(13)
frequency = f0 *(2**(n/12)) #(bigger than 0 so increasing)
# cents = 100*n

plt.plot(n, frequency, marker='o')

for i, f in enumerate(frequency):
    plt.text(i, f, f"{f:.1f}", ha='center', va='bottom')

plt.xlabel("1 step")
plt.ylabel("Frequency (Hz)")
plt.title("12-Tone Equal Temperament")
plt.grid()

plt.show()
```

and from here, cents are just the points between semitone. 1 semitones are 100 cents. So, one cent is $2^{1/1200}$ in this case.

---
***just intonation:*** frequency ratio expressed as fraction between two small integers.
- Instead of dividing an octave evenly, just intonation builds interval based on **natural harmonic relationships.**  

**Example:**
- $C = 256Hz$ Root 
- $E = 256 \times \frac{5}{4} = 320Hz$ (Major third) 
- $G = 256 \times \frac{3}{2} = 384Hz$ (Perfect Fifth)

**Why?**
- These ratios comes from harmonic series. 
- Chords sound very pure.
- No roughness.

---

***Pythagorean tuning:*** every frequency ratio is based on ratio 3:2 (3 limit just tuning).\
**Example:**
- $A = 440Hz$.
- $E = 440 \times \frac{3}{2} = 660Hz \to $ go up to perfect fifth.
- $B = 660 \times \frac{3}{2} \times \frac{1}{2} = 495Hz$ another fifth.
$F^{\sharp} = 495 \times \frac{3}{2} = 742.5Hz$ another fifth. \
