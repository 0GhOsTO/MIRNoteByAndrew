# Chapter 1: Basic Musical Representations and MIR


## Music Information Retrieval

Music Information Retrieval (MIR) studies how computers represent, analyze, and understand music. It sits between music theory, signal processing, and machine learning. I am learning about Music Information Retrieval through the mirdotcom recommended by Professor Shuo Zhang from Tufts.


## Basic Knowledge for Music and Sound

- **Visual form of music** is score and music sheet
- **Pitch** is the high and low of a note; related to frequency
- **Octave** is the interval between two notes. Usually 440 Hz between one octave.
- **Pitch class** is an integer number of octaves apart. Cs like [C1, C2, ...]
- **Equal Temperament** refers to the standard practice of dividing the octave into 12 semitones
- **Key signatures** are the keys (the ones with a bunch of sharps and flats at the front)
- **Time signatures** are the rhythmic structures like 3/4, 6/8 at the front
- **Tempo** notes how slow or fast (BPM)


## Symbolic Representations

- **Piano-Roll** representation refers to anything that looks like a result of piano roll
- **MIDI** — Musical Instrument Digital Interface
  - **MIDI note number**: integer between 0 and 127 that represents the note's pitch
  - **Key velocity**: integer between 0 and 127 that represents the intensity of the sound
  - **MIDI Channel**: integer between 0 and 15 that prompts the synthesizer to use a specific instrument
  - **Clock pulses / ticks**: if the number of pulses per quarter note (PPQN) is defined to 120, then 60 ticks will represent the length of an eighth note
- **MusicXML** — a standard XML-based format for representing Western music notation. Unlike MIDI, it encodes human-readable score information: pitches, rhythms, articulations, and structure.


### About Ticks and Pulses

*(I didn't understand this at first, so here's my breakdown)*

- A quarter note (PPQN = Pulses Per Quarter Note) always equals the defined PPQN. For example, if PPQN = 120, a quarter note = 120 ticks.
- Then: quarter note = 120 × 1, eighth note = 120 ÷ 2, etc.
- **Why ticks?** The timing is relative, so we don't need to calculate absolute time every time.
- **Low PPQN**: the grid is too coarse — it may sound robotic. **High PPQN**: the grid is fine and can capture subtle sounds or imperfections.


### MusicXML Example

The snippet below shows two 16th notes (B♭4 and D5) beamed together inside a measure:

```xml
<measure number="2">
  <note>
    <pitch>
      <step>B</step>
      <alter>-1</alter>    <!-- flat -->
      <octave>4</octave>
    </pitch>
    <duration>1</duration>
    <voice>1</voice>
    <type>16th</type>
    <stem>down</stem>
    <beam number="1">begin</beam>
    <beam number="2">begin</beam>
  </note>
  <note>
    <pitch>
      <step>D</step>
      <octave>5</octave>
    </pitch>
    <duration>1</duration>
    <voice>1</voice>
    <type>16th</type>
    <stem>down</stem>
    <beam number="1">continue</beam>
    <beam number="2">continue</beam>
  </note>
  <!-- ... more notes ... -->
</measure>
```
