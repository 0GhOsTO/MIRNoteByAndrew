## Audio as Signal

Musicians job is to change **sheet music** $\to$ **sound**(air presure & oscillation).
Sound is **longitudinal wave** going through the air. 

**Audio** refers to the production, transmission, and reception of sounds that are audible by humans. **Audio signal** is a representation of sound that represents air fluctuations caused by air vibrations. **Audio Representations** are trying to reproduce acoustic realization of a piece of music.

## Waveform and Time-Domain Representation
Computers can only capture the discrete moments of the time.
**Sampling Frequency** or **Sampling Rate** is the rate at which a computer captures audio data. Usually 44100Hz for this.


```python
import urllib.request

url = "https://raw.githubusercontent.com/musicinformationretrieval/musicinformationretrieval.com/gh-pages/mirdotcom/assets/audio/c_strum.wav"
fp, _ = urllib.request.urlretrieve(url, "c_strum.wav")
x, sr = librosa.load(fp)
ipd.Audio(x, rate=sr)
```

```python
plt.figure(figsize=(15, 5))
librosa.display.waveshow(x, sr=sr, alpha = 0.8)
```

## Timbre: Temporal Indicators
**Timbre** is quality of the tone that distinguishes different instrument. 
**Envelope** of a signal is smooth curve that approximates the amplitude extremes of a waveform over time.

Envelopes are modeled by **Attack-Decay-Sustain-Release (ADSR)** model.
- **Attack**: Time taken for the sound to reach its maximum amplitude after being played.
- **Decay**: Time taken for the sound to decrease from the maximum amplitude to the sustain level.
- **Sustain**: Level of amplitude maintained while the note is held.
- **Release**: Time taken for the sound to fade from the sustain level to silence after the note is released.



## Timbre: Spectral Indicators
**Partials:** dominant frequencies in a sound that contribute to its timbre / lowest partial is called ***fundamental frequency***.




```python
T = 2.0  # seconds
f0 = 1047.0
sr = 22050
t = numpy.linspace(0, T, int(T * sr), endpoint=False)  # time variable
x = 0.1 * numpy.sin(2 * numpy.pi * f0 * t) #------------ need to do this more(syntzx wise).
# t = [0, 1/sr, 2/sr, 3/sr, ...]
# x = amplitude * sin(angular_frequency * time)
ipd.Audio(x, rate=sr)
```
