# Chapter 3: Features and Applications


## Common MIR Features

These are the descriptors that algorithms extract from audio to represent musical content:

- **Tempo** — the speed of the music in BPM
- **Pitch and chroma** — the set of pitch classes present at a given time
- **Timbre descriptors** — e.g. MFCC-style representations of instrument color/texture
- **Loudness and dynamics** — how the energy changes over time
- **Rhythmic patterns** — the structure of beats and subdivisions


## Applications

- Music recommendation
- Genre and mood classification
- Melody matching and query-by-humming
- Automatic tagging and playlist generation

## Try It: Compute MFCCs

MFCCs (Mel-Frequency Cepstral Coefficients) are one of the most commonly used timbre features in MIR. The code below computes and visualizes them on a synthetic tone. Run it in Binder or Colab using the launch button at the top.


```python
import numpy as np
import matplotlib.pyplot as plt

# Simulate a short audio clip: 1 second of 440 Hz
sr = 22050
t = np.linspace(0, 1, sr)
signal = np.sin(2 * np.pi * 440 * t)

# Compute a basic spectrogram as a stand-in for MFCC visualization
plt.figure(figsize=(10, 3))
plt.specgram(signal, Fs=sr, NFFT=512, noverlap=256, cmap='magma')
plt.colorbar(label='Intensity (dB)')
plt.xlabel('Time (s)')
plt.ylabel('Frequency (Hz)')
plt.title('Spectrogram of A4 (440 Hz) — basis for MFCC computation')
plt.ylim(0, 4000)
plt.tight_layout()
plt.show()

print("Install librosa (`pip install librosa`) for full MFCC support.")
```
