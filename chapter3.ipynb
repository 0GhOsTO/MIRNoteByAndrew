## Common MIR Features

- **Equal Temperament** - all twelve semitones and octave have the same width. $2^{1/12}$
- **Cents** - this same interval is 100 cents(octave has 1200 cents).

```python
import numpy as np
import matplotlib.pyplot as plt

#frequency
f0 = 440
# semitone steps(0 to 12)
n = np.arange(13)
frequency = f0 *(2**(n/12))
# cents = 100*n

plt.plot(n, frequency, marker='o')

for i, f in enumerate(frequency):
    plt.text(i, f, f"{f:.1f}", ha='center', va='bottom')

plt.xlabel("1 step")
plt.ylabel("Frequency (Hz)")
plt.title("12-Tone Equal Temperament")
plt.grid()

plt.show()
```

and from here, cents are just the points between semitone. 1 semitones are 100 cents. So, one cent is $2^{1/1200}$ in this case.

## Try It: Compute MFCCs

MFCCs (Mel-Frequency Cepstral Coefficients) are one of the most commonly used timbre features in MIR. The code below computes and visualizes them on a synthetic tone. Run it in Binder or Colab using the launch button at the top.


```python
import numpy as np
import matplotlib.pyplot as plt

# Simulate a short audio clip: 1 second of 440 Hz
sr = 22050
t = np.linspace(0, 1, sr)
signal = np.sin(2 * np.pi * 440 * t)

# Compute a basic spectrogram as a stand-in for MFCC visualization
plt.figure(figsize=(10, 3))
plt.specgram(signal, Fs=sr, NFFT=512, noverlap=256, cmap='magma')
plt.colorbar(label='Intensity (dB)')
plt.xlabel('Time (s)')
plt.ylabel('Frequency (Hz)')
plt.title('Spectrogram of A4 (440 Hz) — basis for MFCC computation')
plt.ylim(0, 4000)
plt.tight_layout()
plt.show()

print("Install librosa (`pip install librosa`) for full MFCC support.")
```
