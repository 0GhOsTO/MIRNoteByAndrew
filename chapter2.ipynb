# Chapter 2: Audio Representations


## Audio as Signal

An audio signal lives in the **time domain** — it is a sequence of amplitude values sampled over time. When you record sound, you are capturing how air pressure changes at a microphone over time.


## Spectrogram Basics

A spectrogram transforms a time-domain audio signal into a 2D representation:

- The **x-axis** is time.
- The **y-axis** is frequency.
- **Color intensity** represents energy at each time-frequency location.

This is computed using the Short-Time Fourier Transform (STFT), which slices the signal into overlapping windows and applies a Fourier transform to each.


## Why This Matters in MIR

Spectrograms make musical structure easier for algorithms to detect, including beat pulses, harmonic patterns, and timbral texture.


## Try It: Load and Play a Real Audio File

Using `mirdotcom` to fetch a sample audio file, then loading it with `librosa` and playing it inline.


```python
import librosa
import IPython.display as ipd
import urllib.request

url = "https://musicinformationretrieval.com/assets/audio/c_strum.wav"
fp, _ = urllib.request.urlretrieve(url, "c_strum.wav")
x, sr = librosa.load(fp)
ipd.Audio(x, rate=sr)
```

## Try It: Plot Its Spectrogram


```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 4))
plt.specgram(x, Fs=sr, NFFT=1024, noverlap=512, cmap='inferno')
plt.colorbar(label='Intensity (dB)')
plt.xlabel('Time (s)')
plt.ylabel('Frequency (Hz)')
plt.title('Spectrogram of c_strum.wav')
plt.tight_layout()
plt.show()
```

> **My takeaway:** Spectrograms turn sound into a visual map that is easier to model computationally.
